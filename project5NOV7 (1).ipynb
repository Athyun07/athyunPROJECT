```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df=pd.read_csv("/content/student_performance_clustering_dataset.csv")
df
```

```python
X=df[["Study_Hours_Per_Day","Attendance_Percentage",	"Assignment_Score"]]
X.head()
```

```python
Kmeans=KMeans(n_clusters=5,random_state=42,n_init=10)
Kmeans.fit(X)
df["Cluster"]=Kmeans.labels_
df.head()
```

```python
plt.scatter(df["Study_Hours_Per_Day"],df["Attendance_Percentage"],df["Assignment_Score"],c=df["Cluster"])
plt.xlabel("Study_Hours_Per_Day")
plt.ylabel("Attendance_Percentage")
plt.title("Student Performance Clustering")
plt.show()
```

```python
print(Kmeans.cluster_centers_)
```
