```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df=pd.read_csv("/content/loan_approval_classification_dataset.csv")

df
```

```python
df.info()
```

```python
df.head()
```

```python
x=df.drop('Loan_Approved',axis=1)
y=df['Loan_Approved']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

```

```python
lr=LogisticRegression(max_iter=200)
lr.fit(x_train,y_train)
lr_pred=lr.predict(x_test)
print("Accuracy:",accuracy_score(y_test,lr_pred))
```

```python
dt=DecisionTreeClassifier(random_state=42)
dt.fit(x_train,y_train)
dt_pred=dt.predict(x_test)
print("Accuracy:",accuracy_score(y_test,dt_pred))
```

```python
knn=KNeighborsClassifier(n_neighbors=3)
knn.fit(x_train,y_train)
knn_pred=knn.predict(x_test)
print("Accuracy:",accuracy_score(y_test,knn_pred))
```

```python
svm=SVC()
svm.fit(x_train,y_train)
svm_pred=svm.predict(x_test)
print("Accuracy:",accuracy_score(y_test,svm_pred))
```

```python
result = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree", "KNN", "SVM"],
    "Accuracy": [accuracy_score(y_test, lr_pred), accuracy_score(y_test, dt_pred), accuracy_score(y_test, knn_pred), accuracy_score(y_test, svm_pred)]
})
result
```
