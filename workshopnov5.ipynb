```python
students=[]
subjects=["python","Mathematics","statistics"]
```

step 2-Add student


```python
def add_student():
  student_id=input("enter student id:")
  #check for dupilicate ID
  for student in students:
    if student in students:
      if student["id"]==student_id:
        print("student ID already exists!")
        return
  name=input("Enter Name:")
  department=input("enter Department")
  marks={}
  for subject in subjects:
    try:
      mark=float(input(f"enter {subject}Marks:"))

      if mark <0 or mark> 100:
        print("Marks must be between 0 and 100")
        return
      marks[subject]=mark
    except ValueError:
      print("please enter a valid number.")
      return
    student={
        "id":student_id,
        "name":name,
        "department":department,
        "marks":marks
    }
    students.append(student)
    print("student added successfully!")
```

```python
def view_student():
  if len(students)==0:
    print("No students found")
    return

  for student in students:
    print(f"ID:{student['id']}")
    print(f"Name:{student['name']}")
    print(f"Department:{student['department']}")
    print("marks:",student["marks"])
```

```python

def search_student():
  search_id=input("enter student ID to search:")
  for student in students:
    if student["id"]==search_id:
      print(f"ID:{student['id']}")
      print(f"Name:{student['name']}")
      print(f"Department:{student['department']}")
      print("marks:",student["marks"])
      return
  print("student not found")
```

```python
def calculate_grade():
  search_id=input("enter student ID to calculate grade:")
  for student in students:
    if student["id"]==search_id:
      total_marks=sum(student["marks"].values())
      average=total_marks/len(student["marks"])
      if average >=90:
        grade="A+"
      elif average >=80:
        grade="A"
      elif average >=70:
        grade="B"
      elif average >=50:
        grade="C"
      elif average >=40:
        grade="D"
      else:
        grade="FAIL"

      print(f"student:",student["name"])
      print("Total Marks",total_marks)
      print("Average:",round(average,2))
      print("Grade:",grade)
      return
  print("student not found!")
```

```python
while True:
  print("\\n====STUDENT MANAGEMENT SYSTEM====")
  print("1.Add Student")
  print("2.View Students")
  print("3.Search Student")
  print("4.Calculate Grade")
  print("5.Exit")
  try:
    choice = int(input("Enter your choice:"))
    if choice == 1:
      add_student()
    elif choice == 2:
      view_student()
    elif choice == 3:
      search_student()
    elif choice == 4:
      calculate_grade()
    elif choice == 5:
      print("Goodbye!")
      break
  except ValueError:
    print("please enter a valid choice")
    continue
```
